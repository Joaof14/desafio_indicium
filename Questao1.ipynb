```python
import pandas as pd
import sqlite3
```

```python

# 1. Carregar o dataset fornecido
df_vendas = pd.read_csv('datasets/vendas_2023_2024.csv')

# 2. Criar uma conexão com um banco de dados SQLite em memória
conn = sqlite3.connect(':memory:')

# 3. Salvar o DataFrame como uma tabela SQL chamada 'vendas'
df_vendas.to_sql('vendas', conn, index=False, if_exists='replace')

print("Tabela 'vendas' estruturada com sucesso para consultas SQL.")
```

### SQL e validação (Questão 1.1 & 1.2)
Query para coletar informaçòes de datas (mínimo e máximo), e valroes (mínimo, máximo e média)

```python
query = """
SELECT 
    COUNT(*) AS total_linhas,
    (SELECT COUNT(*) FROM pragma_table_info('vendas')) AS total_colunas,
    MIN(sale_date) AS data_minima,
    MAX(sale_date) AS data_maxima,
    MIN(total) AS valor_minimo,
    MAX(total) AS valor_maximo,
    AVG(total) AS valor_medio
FROM vendas;
"""

# Executar a query e exibir o resultado
resultado = pd.read_sql(query, conn)
resultado
```

Isso significa que, sem tratamento, o computador não consegue ordenar as datas corretamente, sendo necessário escolher um formato para padroinzar as datas.

Qualidade e Confiabilidade: O dataset apresenta lacunas (IDs de venda pulados por exemplo). Portanto, o diagnóstico é que os dados não são confiáveis para decisão imediata e exigem uma etapa de Engenharia de Dados (limpeza) antes de qualquer aplicação de IA para prever vendas.
