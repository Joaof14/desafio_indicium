```python
# carregando o dataset
df_produtos = pd.read_csv("produtos_raw.csv")
```

### Pontos mais chamativos após análise exploratória
- Temos 150 codigos únicos, mas 157 linhas, indicando duplicatas
- Preço está em tipo string
- Categorias com grafia não padronizada/errada, dificultando agregação.

### Parte 1 - Padronização de categorias

```python
def padronizar_categoria(txt):
    txt = str(txt).upper().replace(" ", "") # Remove espaços e coloca em caixa alta
    if 'ELETR' in txt:
        return 'eletrônicos'
    elif 'PROP' in txt:
        return 'propulsão'
    elif 'ANCOR' in txt or 'ENCOR' in txt:
        return 'ancoragem'
    return 'outros'

df_produtos['actual_category'] = df_produtos['actual_category'].apply(padronizar_categoria)
```

```python
df_produtos['actual_category'].unique()
```

### Parte 2 - Conversão dos valores de preço de tipo string para numérico float

```python
# Remove o prefixo 'R$ ' e converte a coluna price para float
df_produtos['price'] = df_produtos['price'].str.replace('R$ ', '', regex=False).astype(float)
```

```python
df_produtos['price'].info()
```

### Parte 3 - Remoção de Duplicatas

```python
# existem mais linhas que valores únicos em id
df_produtos = df_produtos.drop_duplicates()
```

```python
# Visualizando o resultado final
print("Dados normalizados com sucesso.")
df_produtos.head()
```
