# Questão 3 - Custos de Importação

## Carregar o arquivo JSON e gere um novo arquivo CSV organizando-o de acordo com a definição na imagem abaixo.

```python
import pandas as pd
import json
```

```python
# 1. Carregar o arquivo JSON original
with open('datasets/custos_importacao.json', 'r', encoding='utf-8') as f:
    dados_brutos = json.load(f)
```

```python
# 2. Normalizar os dados usando a chave correta: 'historic_data'

# A função json_normalize vai 'achatar' a lista historic_data 
# e replicar os campos do produto para cada registro de preço.
df_custos = pd.json_normalize(
    dados_brutos, 
    record_path=['historic_data'], 
    meta=['product_id', 'product_name', 'category']
)
df_custos
```

```python
# 3. Reordenar e renomear colunas para o padrão definido
df_custos = df_custos[['product_id', 'product_name', 'category', 'start_date', 'usd_price']]
```

```python
#Análise exploratória rapida
df_custos.info()
```

```python
# 4. Conversão de tipos para garantir a integridade solicitada pela diretoria
df_custos['product_id'] = df_custos['product_id'].astype(int)
df_custos['usd_price'] = df_custos['usd_price'].astype(float)

# Padronizando o formato da data (ex: 10/08/2016 para o padrão ISO YYYY-MM-DD)
df_custos['start_date'] = pd.to_datetime(df_custos['start_date'], dayfirst=True).dt.date
```

```python
# 5. Salvar o resultado final em CSV
df_custos.to_csv('datasets/custos_importacao_normalizado.csv', index=False)
```
